```python
data  = "/content/ipl_matches_data.xlsx"
Ipl_Century = "/content/ipl_centuries.xlsx"
Team_Data = "/content/teams_data.xlsx"
Player_Data = "/content/players-data-updated.xlsx"
Ball_data = "/content/ball_by_ball_data.xlsx"
```

#DATA 1


```python
# ==========================
# Data 1  ipl_matches_data
# ==========================

df = pd.read_excel(data)
df.fillna(0)
df.dropna()
```

```python
df.head()
```

```python
# line plot
wins = df["venue"].value_counts()
plt.figure(figsize=(10,15))
plt.barh(wins.index, wins.values)
plt.title("Matches of Home Team on Single venue")
plt.xticks(rotation=90)
plt.xlabel("Matches")
plt.ylabel("Teams")
plt.show()
```

```python
#Bar plot
# Match Winners

wins = df["match_winner"].value_counts()
plt.figure(figsize=(12,8))
plt.bar(wins.index, wins.values)
plt.xticks(rotation=90)
plt.xlabel("Teams")
plt.ylabel("Wins")
plt.show()
```

```python
wins = df["match_winner"].value_counts()
print("Most Wins By Team: ")
print("Team: ", wins.idxmax())
print("Wins: ",wins.max());

print()
print("Top 5: ")
print(wins.head(5))
```

```python
#pie Chart
#Toss winners

wins = df['toss_winner'].value_counts()
plt.figure(figsize=(8,8))

plt.pie(  wins.values,
    labels=wins.index,
    autopct="%1.1f%%",
    startangle=0)

plt.title("Toss Wins by Team ")
plt.show()
```

```python
#line Plot
season = df['season_id'].value_counts().sort_index()
plt.figure(figsize=(12,6))
plt.plot(season.index, season.values,marker="o", linewidth=1)
plt.xlim(2008, 2026)
plt.xticks(rotation=0)
plt.xlabel("IPL- Seasons Year Wise")
plt.ylabel("Total Matches")
plt.title("Matches per Seasons")
plt.plot()
```

```python
result = df['Result'].value_counts()

plt.pie(result.values,labels=result.index, autopct="%1.1f%%")
plt.title("Result affected by Centuries ")
plt.legend()
plt.show()
```

#DATA 3

```python
# ==========================
# Data 3  Ball_data = "/content/ball_by_ball_data.xlsx"
# ==========================
```

```python
df= pd.read_excel(Ball_data)
df.fillna(0)
```

```python
bowler= df['team_bowling'].value_counts()
plt.figure(figsize=(12,8))
plt.bar(bowler.index , bowler.values)
plt.xticks(rotation=90)
plt.title("Most balls bowled by teams ")
plt.xlabel("Teams")
plt.ylabel("Balls bowled")
plt.plot()
```

```python
Team= df['team_bowling'].value_counts()
print("Most Balls Bowled by Team : ")
print(Team.idxmax())
print(Team.max())
print()
print("Min Balls Bowled by Team : ")
print(Team.idxmin())
print(Team.min())
print()
```

```python
bowler = df['bowler'].value_counts()

print("Most balls Bowled by Bowler: ")
print(bowler.idxmax())
print(bowler.max())
print("===================")
print("Min Balls bowled by bowler: ")
print(bowler.idxmin())
print(bowler.min())
print("===================")

# MOST BALLS BOWLED BY BOWLER
top = df['bowler'].value_counts().head(10)
plt.figure(figsize=(7,5))
plt.barh(top.index,top.values)
plt.xlabel("Balls Bowled ")
plt.ylabel("Bowler Name ")
plt.title("MOST BALLS BOWLED BY BOWLER")
plt.tight_layout()
plt.plot()

# MIN BALLS BOWLED BY BOWLER
low = df['bowler'].value_counts().tail(10)
plt.figure(figsize=(7,5))
plt.barh(low.index,low.values)
plt.yticks(rotation=30)
plt.xlabel("Balls Bowled ")
plt.ylabel("Bowler Name ")
plt.title("MIN BALLS BOWLED BY BOWLER")
plt.tight_layout()
plt.plot()

```

```python
team = df['team_batting'].value_counts()
plt.figure(figsize=(12,8))
plt.plot(team.index,team.values,marker='o')
plt.xticks(rotation=90)
plt.xlabel("Teams ")
plt.ylabel("Balls faced")
plt.title("Most Balls Faced By teams ")
plt.plot()
```

```python
balls = df['team_batting'].value_counts()
print("most balls faced by team:")
print(balls.idxmax())
print(balls.max())
print()
print("min balls faced by team:")
print(balls.idxmin())
print(balls.min())
print()
print(balls.head())
```

```python
# MOST BALLS FACED BY BATSMAN
bats = df['batter'].value_counts()
print("Most Balls Faced By Batsman: ")
print(bats.idxmax())
print(bats.max())
print("========================")
print("Min Balls Faced By Batsman: ")
print(bats.idxmin())
print(bats.min())
print("========================")

# TOP 10 Batsman , Most Balls Faced
top = df['batter'].value_counts().head(10)
plt.figure(figsize=(7,5))
plt.bar(top.index,top.values)
plt.xticks(rotation=90)
plt.title("Most Balls Faced By Batsman: ")
plt.xlabel("Balls Faced")
plt.ylabel("Batsman")
plt.tight_layout()
plt.plot()

# TOP 10 Batsman , Min  Balls Faced
low = df['batter'].value_counts().tail(10)
plt.figure(figsize=(7,5))
plt.bar(low.index,low.values)
plt.xticks(rotation=90)
plt.title("Min Balls Faced By Batsman: ")
plt.xlabel("Balls Faced")
plt.ylabel("Batsman")
plt.tight_layout()
plt.plot()
```

```python
#line Plot
season = df['season_id'].value_counts().sort_index()
print("Max Balls Bowled In season: ")
print(season.idxmax())
print(season.max())
print("===========================")
print("Min Balls Bowled In season: ")
print(season.idxmin())
print(season.min())
print("===========================")
plt.figure(figsize=(12,6))
plt.plot(season.index, season.values,marker="o", linewidth=1)
plt.xlim(2008, 2026)
plt.xticks(rotation=0)
plt.xlabel("IPL- Seasons Year Wise")
plt.ylabel("Total Balls Bowled ")
plt.title("Balls Bowled per Seasons")
plt.plot()

```

```python
# PIE CHART
bat = df["batsman_type"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(bat.values,labels=bat.index,autopct="%1.1f%%")
plt.title("Batsman Type")
plt.tight_layout()
plt.legend()
plt.plot()
```

```python
# PIE CHART
ball = df["bowler_type"].value_counts()
plt.figure(figsize=(12,8))
plt.barh(ball.index,ball.values)
plt.xticks(rotation=90)
plt.title("Most Balls Bowled By Bowlers Type")
plt.ylabel("Bowlers type")
plt.xlabel("Balls Bowled")
plt.tight_layout()

plt.plot()
```
